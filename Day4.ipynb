```python
from sentence_transformers import SentenceTransformer
import chromadb
client = chromadb.Client()
```

```python
model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
collection = client.create_collection(
    name="document",
    get_or_create=True
)
document = [
    "Employees receive 20 days of annual leave.",
    "Employees can work remotely up to 3 days per week.",
    "Leave requests must be submitted at least 2 days in advance.",
    "Employees must be available between 10 AM and 5 PM while working remotely."
]
ids = ["doc1", "doc2", "doc3", "doc4"]
metadatas = [
    {"source": "company_policy.txt"},
    {"source": "company_policy.txt"},
    {"source": "company_policy.txt"},
    {"source": "company_policy.txt"}
]

collection.add(
    documents=document,
    ids=ids,
    metadatas=metadatas
)
```

```python
text = "Machine learning is a method of learning patterns from data."

embedding = model.encode(text)

print(embedding)
print(len(embedding))
```

```python
texts = [
    "Machine learning algorithms learn patterns from data.",
    "Artificial intelligence models can learn from examples.",
    "I like eating pizza."
]
```

```python
embeddings = model.encode(texts)

print(embeddings.shape)
```

```python
from sentence_transformers.util import cos_sim

similarity = cos_sim(embeddings, embeddings)

print(similarity)
```

```python
result = collection.get()

print(result)
```

```python
query = "How many vacation days do employees get?"
results = collection.query(
    query_texts=[query],
    n_results=2
)
```

```python
print(results)
```

```python
!pip install -q transformers accelerate
from transformers import pipeline
```

```python
generator = pipeline(
    "text-generation",
    model="Qwen/Qwen2.5-0.5B-Instruct",
    device_map="auto"
)
```

```python
prompt = """
Explain what machine learning is in simple terms.
"""

result = generator(
    prompt,
    max_new_tokens=100
)

print(result[0]["generated_text"])
```

```python
context = "\n".join(results["documents"][0])

print(context)
```

```python
prompt = f"""
You are a helpful assistant.

Answer the question using only the context provided below.

Context:
{context}

Question:
{query}

Answer:
"""
```

```python
result = generator(
    prompt,
    max_new_tokens=100
)

print(result[0]["generated_text"])
```
