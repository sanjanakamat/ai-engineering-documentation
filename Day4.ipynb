```python
def generate_answer(query, documents):

    context = "\n".join(documents)

    prompt = f"""
You are a helpful assistant.

Answer the question using ONLY the provided context.

If the answer is not present in the context,
say "I don't have enough information to answer this."

Do not make up information.
Context:
{context}

Question:
{query}

Answer:
"""

    result = generator(
        prompt,
        max_new_tokens=100
    )

    return result[0]["generated_text"]
```

```python
from pypdf import PdfReader

reader = PdfReader("company_data.pdf")

print("Pages:", len(reader.pages))
```

```python
text = ""

for page in reader.pages:
    page_text = page.extract_text()


    if page_text:
        text += page_text + "\n"

print(text[:2000])
```

```python
chunk_size = 1000 #naive chunking strategy is not efficient,we might seperate related information

chunks = []

for i in range(0, len(text), chunk_size):
    chunks.append(text[i:i + chunk_size])

print("Number of chunks:", len(chunks))
```

```python
#chunk overlapping-helps to preserve the context
chunk_size = 1000
overlap = 200

chunks = []

start = 0

while start < len(text):

    end = start + chunk_size

    chunk = text[start:end]

    chunks.append(chunk)

    start = end - overlap
```

```python
#recursive text splitting using Langhain
!pip install -q langchain-text-splitters
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=200
)
```

```python
chunks = splitter.split_text(text)

print("Number of chunks:", len(chunks))
```

```python
def retrieve_documents(query, k=3):

    results = collection.query(
        query_texts=[query],
        n_results=k,
        include=["documents", "metadatas", "distances"]
    )

    retrieved = []

    for i in range(len(results["documents"][0])):

        retrieved.append({
            "text": results["documents"][0][i],
            "metadata": results["metadatas"][0][i],
            "distance": results["distances"][0][i]
        })

    return retrieved
```

```python
docs = retrieve_documents(
    "How many days of annual leave do employees get?"
)
```

```python
for doc in docs:
    print(doc)
```

```python
context = ""

for doc in docs:

    context += f"""
Source: {doc["metadata"]["source"]}

Content:
{doc["text"]}

"""
```

```python
query = "What is the company's policy on space travel?"
docs = retrieve_documents(query, k=3)

for doc in docs:
    print("=" * 50)
    print("Distance:", doc["distance"])
    print(doc["text"])
```

```python
MAX_DISTANCE = 1.5
relevant_docs = [
    doc for doc in docs
    if doc["distance"] <= MAX_DISTANCE
]
if not relevant_docs:
    print("No sufficiently relevant information found.")
```

```python
queries = [
    "What is the annual leave policy?",
    "What are the remote work rules?",
    "Who is eligible for the policy?",
    "What is the company's policy on space travel?"
]

for query in queries:

    print("\n\nQUERY:", query)

    docs = retrieve_documents(query, k=3)

    for doc in docs:
        print(
            "Distance:",
            round(doc["distance"], 4),
            "| Source:",
            doc["metadata"]
        )
```

```python
from sentence_transformers import CrossEncoder
reranker = CrossEncoder(
    "cross-encoder/ms-marco-MiniLM-L-6-v2"
)
```

```python
query = "What are the remote work requirements?"
documents = [
    "Employees receive 20 days of annual leave.",
    "Employees can work remotely up to 3 days per week.",
    "Employees working remotely must be available between 10 AM and 5 PM.",
    "Employees must submit leave requests two days in advance."
]
pairs = [
    [query, document]
    for document in documents
]

scores = reranker.predict(pairs)

print(scores)

```

```python
ranked = sorted(
    zip(documents, scores),
    key=lambda x: x[1],
    reverse=True
)
for document, score in ranked:
    print("Score:", score)
    print(document)
    print()
```

```python
query = "What are the remote work requirements?"

results = collection.query(
    query_texts=[query],
    n_results=10,
    include=["documents", "metadatas", "distances"]
)
documents = results["documents"][0]
metadatas = results["metadatas"][0]

pairs = [
    [query, document]
    for document in documents
]
scores = reranker.predict(pairs)
```

```python
ranked_results = []

for document, metadata, score in zip(
    documents,
    metadatas,
    scores
):
    ranked_results.append({
        "document": document,
        "metadata": metadata,
        "score": float(score)
    })
```

```python
ranked_results.sort(
    key=lambda x: x["score"],
    reverse=True
)
```

```python
for result in ranked_results:
    print("=" * 60)
    print("Score:", result["score"])
    print("Metadata:", result["metadata"])
    print(result["document"])
```
