```python
x=20
def outer():
    x = 10 # This 'x' is now local to 'outer' and shadows the global 'x'
    def inner():
        nonlocal x # This 'x' correctly refers to the 'x' in 'outer'
        x += 5

    inner()
    print(x)
print(x)
outer()
```

```python
def largeVals():
  for i in range(1000):
    yield i*i
num=largeVals()
for n in num:
  print(n)
```

```python
def get_squares(n):
    for i in range(n):
        yield i * i      # ← hands out ONE value, then pauses

for sq in get_squares(1_000_000):
    print(sq)
```

```python
class Example:
  count=0
  def __init__(self,name,age):
    self.name=name
    self.age=age
  @classmethod
  def clsmeth(cls,name,age):
    cls.count+=1

d=Example.clsmeth("sanjana",20)
print(Example.count)
e=Example.clsmeth("ansh",30)
print(Example.count)

```

```python
#inheritence
class A:
  def hello(self):
    print("A")
class B:
  def hello(self):
    print("B")
class c(A,B):
  pass
c=c()
c.hello()
```

```python
#automatic generation of init,repr and eq through @dataclass
from dataclasses import dataclass
@dataclass
class Point:
  x:int
  y:int
p=Point(1,2)
print(p)
print(p==Point(1,2))
```

```python
#abstract methods in python
from abc import ABC, abstractmethod
class Animal(ABC):
  @abstractmethod
  def speak(self):
    pass
class Dog(Animal):
  def speak(self):
    return "Woof"
d=Dog()
print(d.speak())
```
